```python
import cv2
import threading

# Adresy IP twoich telefonów z DroidCam (sprawdź w aplikacji na telefonie)
url_camera1 = "http://192.168.137.76:4747/video"
url_camera2 = "http://192.168.137.234:4747/video"

# Inicjalizacja przechwytywania
cap1 = cv2.VideoCapture(url_camera1)
cap2 = cv2.VideoCapture(url_camera2)

# Ustawienie bufora na minimalny, aby zmniejszyć opóźnienia (dla IP cam)
cap1.set(cv2.CAP_PROP_BUFFERSIZE, 1)
cap2.set(cv2.CAP_PROP_BUFFERSIZE, 1)

while True:
    # Pobieranie klatek
    ret1, frame1 = cap1.read()
    ret2, frame2 = cap2.read()

    if not ret1 or not ret2:
        print("Błąd pobierania klatki z jednej z kamer.")
        break

    # --- TU WSTAWIASZ SWÓJ KOD PRZETWARZAJĄCY ---
    # Np. zmiana na czarno-białe:
    # frame1 = cv2.cvtColor(frame1, cv2.COLOR_BGR2GRAY)

    # Wyświetlanie (dla testu zmniejszamy okna, żeby się zmieściły)
    cv2.imshow('Kamera 1', cv2.resize(frame1, (640, 480)))
    cv2.imshow('Kamera 2', cv2.resize(frame2, (640, 480)))

    # Wyjście po naciśnięciu 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap1.release()
cap2.release()
cv2.destroyAllWindows()
```

```python
import cv2
import mediapipe as mp
import numpy as np

mp_drawing = mp.solutions.drawing_utils
mp_pose = mp.solutions.pose

def uruchom_detekcje_ruchu(stream_source=1):

    cap = cv2.VideoCapture(stream_source)
    if not cap.isOpened():
        print(f"Błąd: Nie można otworzyć źródła wideo: {stream_source}")
        return

    with mp_pose.Pose(min_detection_confidence=0.5, min_tracking_confidence=0.5) as pose:

        while cap.isOpened():
            ret, ramka = cap.read()
            if not ret:
                break

            ramka = cv2.flip(ramka, 1)

            ramka_rgb = cv2.cvtColor(ramka, cv2.COLOR_BGR2RGB)

            ramka_rgb.flags.writeable = False

            wyniki = pose.process(ramka_rgb)

            ramka_rgb.flags.writeable = True
            ramka = cv2.cvtColor(ramka_rgb, cv2.COLOR_RGB2BGR)

            komunikat_analizy = "Oczekiwanie na detekcję..."

            mp_drawing.draw_landmarks(
                ramka, wyniki.pose_landmarks, mp_pose.POSE_CONNECTIONS,
                mp_drawing.DrawingSpec(color=(245,117,66), thickness=2, circle_radius=2),
                mp_drawing.DrawingSpec(color=(245,66,230), thickness=2, circle_radius=2)
            )

            cv2.imshow('Trener AI - Detekcja Ruchu', ramka)

            if cv2.waitKey(10) & 0xFF == ord('q'):
                break

    cap.release()
    cv2.destroyAllWindows()


if __name__ == '__main__':
    uruchom_detekcje_ruchu(stream_source=1)
```
